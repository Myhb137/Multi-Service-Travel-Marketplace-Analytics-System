```python
is_canceled_corr = num_data.corr()['is_canceled'] 
adr_corr = num_data.corr()['adr']
```

```python
sns.heatmap(is_canceled_corr.to_frame(), cmap='coolwarm', annot=True)
```

```python
for i in is_canceled_corr.index:
    if abs(is_canceled_corr[i]) > 0.1 and i != 'is_canceled':
        print(f"{i} has a correlation of {is_canceled_corr[i]:.2f} with is_canceled.")
```

```python
for i in adr_corr.index:
    if abs(adr_corr[i]) > 0.1 and i != 'adr':
        print(f"{i} has a correlation of {adr_corr[i]:.2f} with adr.")
```

```python
plt.figure(figsize=(15,8))
sns.lineplot(x='arrival_date_month', y='adr', data=data) 
plt.xticks(rotation=45)
plt.title('Average Daily Rate by Arrival Month')    
```

```python
adr_corr_list = ['arrival_date_year' , 'adults', 'children', 'is_repeated_guest','total_of_special_requests']
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

scaler = MinMaxScaler()

df_norm = data.copy()
df_norm[adr_corr_list + ['adr']] = scaler.fit_transform(
    data[adr_corr_list + ['adr']]
)

plt.figure(figsize=(10,6))

for col in adr_corr_list:
    sns.lineplot(x=df_norm[col], y=df_norm['adr'], label=col)

plt.title("ADR vs Features")
plt.legend()
plt.show()
```
